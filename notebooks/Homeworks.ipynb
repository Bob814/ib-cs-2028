## Home 1: Demo

```python
import math
v = float(input("State the initial speed [ms^-1]: "))
angle = float(input("State the launch angle [degrees]: "))
h = float(input("State the initial height [m]: "))
dt = float(input("State the time interval between calculations [s]: "))
g = 9.81

# What I've decided to do is calculate each ds for every dt, so every time dt passes, ill update the
# velocity and estimate the distance traveled in this time dt, then ill add this to the initial height value
# the process will end when velocity is less than or equal to 0, once it is equal to zero, the new value will be
# the max height (MH), if its less than 0, then the previous value will be named MH (this is for the y value of course)
angle_rad = math.radians(angle)
vx = v * math.cos(angle_rad)
vy = v * math.sin(angle_rad)
x = 0.0
y = h
time = 0.0
MH = h
print("the initial coordinates are: (" + str(0) + ";" + str(h) + ")")
while vy > 0:
    prev_y = y
    vy_next = vy - g*dt
    ds = ((vy + vy_next) / 2)*dt

    y = y + ds
    x = x + vx*dt
    time = time + dt
    vy = vy_next
    if vy == 0:
        MH = y
    elif vy < 0:
        MH = prev_y


print("Max height: " + str(MH))
print("Time taken: " + str(time))
print("Horizontal distance: " + str(x))
```

### Sami : Projectile

```python
import math

v = float(input("Input the velocity: "))
angle = float(input("Input the angle: "))
x = float(input(
    "Input the horizontal position [horizontal position cannot be negative]: "))
dt = float(input("Input the difference between measurements: "))
initial_x = float(x)
g = 9.81
XSize = 20
Terrain = [0, 0.5, 1.5, 4, 3.5, 3, 2]
while x < 0:
    x = float(input("Invalid value for x, please input a positive number: "))


def getgroundheight(x):
    i = int(x / XSize)
    if i >= len(Terrain):
        return Terrain[-1]
    else:
        return Terrain[i]


groundheight = getgroundheight(x)
y = groundheight
angle_rad = math.radians(angle)
vy = v * math.sin(angle_rad)
vx = v * math.cos(angle_rad)

while y >= groundheight:
    x = x + vx * dt
    y = y + vy * dt
    vy = vy - g * dt
    groundheight = getgroundheight(x)

Range = x - initial_x
print(f"The distance covered is: {Range:.3f}")
```

### Sami: Tic Tac Toe

```python
from operator import truediv

table = [
    [0, 0, 0],
    [0, 0, 0],
    [0, 0, 0],
]
win = False


def print_Table(table):
    for row in table:
        print(row)


def check_win(table):
    for row in range(3):
        vertical_win_condition = str(
            table[row][0]) + str(table[row][1]) + str(table[row][2])
        if vertical_win_condition == "XXX" or vertical_win_condition == "○○○":
            return True
        else:
            return False

    for column in range(3):
        horizontal_win_condition = table[0][column] + \
            table[1][column] + table[2][column]
        if horizontal_win_condition == "XXX" or horizontal_win_condition == "○○○":
            return True
        else:
            return False

    diag1 = str(table[0][0]) + str(table[1][1]) + str(table[2][2])
    diag2 = str(table[0][2]) + str(table[1][1]) + str(table[2][0])
    if diag1 in ("XXX", "○○○") or diag2 in ("XXX", "○○○"):
        print("Winner by diagonal connection!")
        return True

    return False


while win is False:
    Mark_1 = input(
        "Player_1, enter the row and column of your mark: ").split(",")
    row = int(Mark_1[0]) - 1
    column = int(Mark_1[1]) - 1
    if table[row][column] == 0:
        table[row][column] = "X"
    else:
        while table[row][column] != 0:
            Mark_1 = input(
                "That spot is already taken! Choose another spot: ").split(",")
            row = int(Mark_1[0]) - 1
            column = int(Mark_1[1]) - 1
        table[row][column] = "X"

    print_Table(table)
    if check_win(table):
        win = True
        print("Winner is Player_1")
        break

    Mark_2 = input(
        "Player_2, enter the row and column of your mark: ").split(",")
    row = int(Mark_2[0]) - 1
    column = int(Mark_2[1]) - 1
    if table[row][column] == 0:
        table[row][column] = "○"
    else:
        while table[row][column] != 0:
            Mark_2 = input(
                "That spot is already taken! Choose another spot: ").split(",")
            row = int(Mark_2[0]) - 1
            column = int(Mark_2[1]) - 1
        table[row][column] = "○"
    print_Table(table)
    if check_win(table):
        win = True
        print("Winner is Player_1")
        break
```
